# Allergen

```python
gold_standard = jupyter_helper_functions.load_dataset(
    "../datasets/parker_datasets/allergen/allergen_cleaned_gold_first1000.csv"
)

original = jupyter_helper_functions.load_dataset(
    "../datasets/parker_datasets/allergen/allergen_corrupted_first1000.csv"
)

results_path = "../analysis/results/allergen"

files = glob.glob("../analysis/repairs/allergen/*.csv")
print(files)
primary_key = "code"
```

```python
for file in files:
    cleaned = jupyter_helper_functions.load_dataset(file)
    file_name = os.path.splitext(os.path.basename(file))[0]
    file_name = jupyter_helper_functions.sanitize_filename(file_name)
    print(os.path.join(results_path, f"{file_name}_micro.json"))
    micro_stats = evaluate_dataset_micro(
        gold_standard=gold_standard,
        cleaned_dataset=cleaned,
        original_dataset=original,
        primary_key=primary_key,
    )
    jupyter_helper_functions.save_json(micro_stats, os.path.join(results_path, f"{file_name}_micro.json"), overwrite=True)
    macro_stats = evaluate_dataset_macro(
        gold_standard=gold_standard,
        cleaned_dataset=cleaned,
        original_dataset=original,
        primary_key=primary_key,
    )
    jupyter_helper_functions.save_json(macro_stats, os.path.join(results_path, f"{file_name}_macro.json"), overwrite=True)
```
